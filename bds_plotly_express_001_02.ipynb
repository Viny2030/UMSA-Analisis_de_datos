```python
import plotly.express as px
```

<p align="justify">
Para poder graficar, vamos a utilizar un conjunto de datos correspondientes a un Fondo Común de Inversión</p>

```python
import pandas as pd
import numpy as np
fci = pd.read_csv("https://raw.githubusercontent.com/Viny2030/UMSA-Analisis_de_datos/refs/heads/main/fci.csv")
fci.head()
```

```python
fci.to_csv("fci.csv", index=False)
```

```python
fci.info()
```

```python
fci['Fecha'] = pd.to_datetime(fci['Fecha'], dayfirst=True) # de esta forma dejamos año mes dia, sin hora...
fci.head(3)
```

```python
fci.info()
```

```python
fci = fci.round(2)
fci.head()
```

 # **<font color="DeepPink">Desarrollando gráficos lineales</font>**

<p align="justify">
👀 Con <code>px.line()</code> cada punto de datos se representa como coordenadas (cuya ubicación proviene de la columna $x$ y la columna $y$) generando una línea que une esas coordenadas, en un espacio bidimensional.<br><br> Antes de graficar, vemos los datos de un fondo común de inversión, filtrando por los rescates y las suscripciones:</p>

```python
fci.head(3)
```

```python
fci.Concepto.unique()
```

```python
fci.query("Concepto == 'Rescate'").head()
```

```python
fci.query("Concepto == 'Suscripción'").head()
```

<p align="justify">
Entonces los gráficos líneales de <code>Plotly</code> se implementan como diagramas de dispersión conectados, lo que significa que los puntos se trazan y se conectan con líneas en el orden en que se proporcionan, sin reordenación automática.<br><br>Esto hace posible hacer gráficos como el siguiente (el que se encuentra ordenado por fechas), pero también significa que puede ser necesario ordenar explícitamente los datos antes de graficarlos para evitar que las líneas se muevan "hacia atrás" a través del gráfico.
</p>



<p align="justify">
👀 En el siguiente gráfico, usamos algunos parámetros, tales como:</p>

- el dataframe.
- la serie asignada a $x$ (tambien puede ser una lista de series).
- la serie asignada a $y$.
- el título del gráfico.
- el template utilizado.
- los marcadores sobre la línea.
- y los labels (para cambiar los nombres de las etiquetas).

```python
fig = px.line(fci.query("Concepto =='Rescate'"),
              x=fci.query("Concepto =='Rescate'").Fecha.dt.date,
              y="Valor_de_Cuota",
              title='Fondo Común de Inversión - Rescate',
              template="gridon",
              markers=True,  ##e ordena al gráfico que dibuje un punto físico (marcador) en cada una de las coordenadas exactas de tus datos.
              labels={"Valor_de_Cuota":"Valor de Cuota","x":"Fecha"})
fig.show()
```

<p align="justify">
👀 En el siguiente gráfico es lo mismo, pero se aplica otro filtro, para visualizar solo las suscripciones. Interesante destacar que se puede colocar dentro del parametro dataframe, un filtro del mismo.</p>


```python
fig = px.line(fci.query("Concepto =='Suscripción'"),
              x=fci.query("Concepto =='Suscripción'").Fecha.dt.date,
              y="Valor_de_Cuota",
              title='Fondo Común de Inversión - Suscripción',
              template="gridon",
              markers=True,  ##e ordena al gráfico que dibuje un punto físico (marcador) en cada una de las coordenadas exactas de tus datos.
              labels={"Valor_de_Cuota":"Valor de Cuota","x":"Fecha"})
fig.show()
```

<p align="justify">
👀 En el siguiente gráfico excluye uno de los valores de la serie <code>Concepto</code>, y agrega un parametro denominado <code>color</code>, que incorpora una variable más a nuestro gráfico.</p>


```python
fci.query("Concepto !='Saldo Anterior'")  ## agrupa excluye Saldo Anterior
```

```python
px.line(fci.query("Concepto !='Saldo Anterior'"),
        x=fci.query("Concepto !='Saldo Anterior'").Fecha.dt.date,
        y="Valor_de_Cuota",
        color="Concepto",
        title='Fondo Común de Inversión - Rescate vs. Suscripción',
        template="gridon",
        markers=True,## para activar los puntos o marcadores visuales en cada una de las coordenadas de datos.
        labels={"Valor_de_Cuota":"Valor de Cuota", "x":"Fecha"})
```

<p align="justify">
👀 Ahora cambiamos el parametro <code>markers</code> por <code>symbol</code>, que agrega sobre la línea, distintos símbolos donde estaban los marcadores.</p>


```python
px.line(fci.query("Concepto !='Saldo Anterior'"),
        x=fci.query("Concepto !='Saldo Anterior'").Fecha.dt.date,
        y="Valor_de_Cuota",
        color="Concepto",
        hover_name="Concepto",               # titulo de la etiqueta
        title='Fondo Común de Inversión',
        template="gridon",
        symbol="Concepto",
        labels={"Valor_de_Cuota":"Valor de Cuota", "x":"Fecha"})
```

<p align="justify">
👀 Optimizando un poco mas el gráfico, utilizamos <code>update_traces</code> para colocar sobre la línea, el valor de cada punto. Hay varias posiciones del texto, en este caso, se establece que los valores se ubican por debajo de los puntos de la línea, y a la derecha.</p>


```python
fig = px.line(fci.query("Concepto =='Suscripción'"),
              x=fci.query("Concepto =='Suscripción'").Fecha.dt.date,
              y="Valor_de_Cuota",
              title='Fondo Común de Inversión',
              template="gridon",
              markers=True,
              labels={"Valor_de_Cuota":"Valor de Cuota","x":"Fecha"},
              text="Valor_de_Cuota")

fig.update_traces(textposition="top left")
##le estás ordenando al gráfico que desplace las etiquetas para que queden alineadas arriba y a la izquierda de cada marcador o elemento visual.
fig.show()
```

<p align="justify">
👀 Ahora vamos adicionar botones para poder segmentar nuestro gráfico, para ello hacemos un <code>update_xaxes()</code> con el parámetro <code>rangeselector</code> armando una lista para los botones, y cada boton es un diccionario.</p>


```python
fig = px.line(fci.query("Concepto =='Suscripción'"),
              x=fci.query("Concepto =='Suscripción'").Fecha.dt.date,
              y="Valor_de_Cuota",
              title='Fondo Común de Inversión',
              template="gridon",
              markers=True,
              labels={"Valor_de_Cuota":"Valor de Cuota","x":"Fecha"},
              text="Valor_de_Cuota")

fig.update_traces(textposition="bottom right")

fig.update_xaxes(rangeselector=dict(buttons=list([ ####Modifica las propiedades del eje horizontal (Eje X).
    dict(count=7, label="7 dias", step="day", stepmode="backward"),####Es la cantidad de unidades de tiempo que se van a medir,
    dict(count=14, label="14 dias", step="day", stepmode="backward"),## label etiqueta
    dict(count=1, label="1 mes", step="month", stepmode="backward"), #### Es la unidad de medida de tiempo
    dict(count=2, label="2 meses", step="month", stepmode="backward"), ## Es la dirección del filtro. Significa "hacia atrás" en el tiempo
    dict(label = "Todo el período", step="all")])))

fig.show()
```

```python
fci1 = fci.query("Concepto =='Suscripción'")
fci1.head()
```

```python
# Gráfico de línea
fig = px.line(fci1,
              x=fci1.Fecha.dt.date,
              y="Valor_de_Cuota",
              title='Fondo Común de Inversión',
              template="gridon",
              markers=True,
              labels={"Valor_de_Cuota": "Valor de Cuota", "x": "Fecha"},
              text="Valor_de_Cuota")

# Línea horizontal para la media del Valor_de_Cuota
fig.add_hline(y=fci1.Valor_de_Cuota.mean(),
              line_color="green",
              line_dash="dot",
              )

# Texto
fig.update_traces(textposition="top left")

# Anotacion Valor medio
fig.update_layout(
    annotations=[
        dict(
            x=fci1.Fecha.dt.date.max(),         # Posición horizontal de la anotación
            y=fci1.Valor_de_Cuota.mean(),       # Posición vertical de la anotación
            xref="x",
            yref="y",
            text="Valor medio",
            showarrow=True,
            arrowhead=2,
            ax=0,
            ay=-25,
            font=dict(size=12, color="green"))])

# Selección de rango de fechas en el eje x
fig.update_xaxes(rangeselector=dict(buttons=list([  ##Modifica las propiedades del eje horizontal (Eje X).
    dict(count=7, label="7 días", step="day", stepmode="backward"),##Es la cantidad de unidades de tiempo que se van a medir,
    dict(count=14, label="14 días", step="day", stepmode="backward"),## label etiqueta
    dict(count=1, label="1 mes", step="month", stepmode="backward"), ## Es la unidad de medida de tiempo
    dict(count=2, label="2 meses", step="month", stepmode="backward"),##Es la dirección del filtro. Significa "hacia atrás" en el tiempo
    dict(label="Todo el período", step="all")
])))

fig.show()
```

<p align="justify">
👀 Ahora agregamos una línea constante, con la media de los valores de suscripción, para ello utilizamos <code>add_hline</code> que es una linea horizontal.</p>


```python
# Gráfico de línea
fig = px.line(fci1,
              x=fci1.Fecha.dt.date,
              y="Valor_de_Cuota",
              title='Fondo Común de Inversión',
              template="gridon",
              markers=True,
              labels={"Valor_de_Cuota": "Valor de Cuota", "x": "Fecha"},
              text="Valor_de_Cuota")

# Línea horizontal para la media del Valor_de_Cuota
fig.add_hline(y=fci1.Valor_de_Cuota.mean(),
              line_color="green",
              line_dash="dot",
              )

# Texto
fig.update_traces(textposition="top left")

# Anotacion Valor medio
fig.update_layout(
    annotations=[
        dict(
            x=fci1.Fecha.dt.date.max(),         # Posición horizontal de la anotación
            y=fci1.Valor_de_Cuota.mean(),       # Posición vertical de la anotación
            xref="x",
            yref="y",
            text="Valor medio",
            showarrow=True,
            arrowhead=2,
            ax=0,
            ay=-25,
            font=dict(size=12, color="green"))])

# Selección de rango de fechas en el eje x
fig.update_xaxes(rangeselector=dict(buttons=list([  ##Modifica las propiedades del eje horizontal (Eje X).
    dict(count=7, label="7 días", step="day", stepmode="backward"),##Es la cantidad de unidades de tiempo que se van a medir,
    dict(count=14, label="14 días", step="day", stepmode="backward"),## label etiqueta
    dict(count=1, label="1 mes", step="month", stepmode="backward"), ## Es la unidad de medida de tiempo
    dict(count=2, label="2 meses", step="month", stepmode="backward"),##Es la dirección del filtro. Significa "hacia atrás" en el tiempo
    dict(label="Todo el período", step="all")
])))

fig.show()
```

<p align="justify">
👀 Y verificamos el valor de la media, calculando con <code>mean()</code> para ver si la linea que adicionamos esta bien posicionada, y vemos que si.<br><br>El truco en este último grafico fue filtrar por fuera del gráfico, asignar el resultante de ese filtrado a un objeto, y trabajar con ese objeto en el script del gráfico</p>


```python
fci1.Valor_de_Cuota.mean().round(2)
```
