 # **<font color="#07a8ed">Ejercitando Pandas y Plotly 🐼 </font>**

```python
import numpy as np
import pandas as pd
```

 # **<font color="#07a8ed">Dataframe</font>**

 ## **<font color="#07a8ed">Creando el Dataframe</font>**

```python
url = "https://infra.datos.gob.ar/catalog/sspm/dataset/168/distribution/168.1/download/datos-tipo-cambio-usd-futuro-dolar-frecuencia-diaria.csv"
```

```python
# Crear el dataframe
df = pd.read_csv(url)
```

```python
# Ver los cinco primeros registros
df.head()
```

 ## **<font color="#07a8ed">Trabajando con el Dataframe</font>**

```python
# Crear una columna denominada año, con el año de la columna indice_tiempo
df["año"] = df["indice_tiempo"].str[:4]
```

```python
# Filtrar para el año 2024
df_2024 = df[df["año"] == "2024"]
```

```python
# Filtrar los primeros 60 registros
df_2024 = df_2024.head(60)
```

```python
# Ver la dimensión del dataframe df_2024
df_2024.shape
```

```python
# Dejar las columnas indice_tiempo y tipo_cambio_a3500
df_2024 = df_2024[["indice_tiempo", "tipo_cambio_a3500"]]
```

```python
# Cambiar el nombre de las columnas por fecha y tipo de cambio
df_2024.columns = ["fecha", "tipo_cambio"]
```

```python
# Graficar con Plotly un grafico de lineas, colocar un titulo acorde
import plotly.express as px
fig = px.line(df_2024, x="fecha", y="tipo_cambio", title="Tipo de cambio")
fig.show()
```

```python
# Colocar marcadores sobre la linea del gráfico
fig = px.line(df_2024, x="fecha", y="tipo_cambio", title="Tipo de cambio", markers=True)
fig.show()
```

```python
# Filtrar en el gráfico, los primeros 30 datos
fig = px.line(df_2024.head(30), x="fecha", y="tipo_cambio", title="Tipo de cambio", markers=True)
fig.show()
```

```python
# Agregar al grafico una linea con el valor medio del tipo de cambio
fig = px.line(df_2024, x="fecha", y="tipo_cambio", title="Tipo de cambio", markers=True)
fig.add_hline(y=df_2024["tipo_cambio"].mean())
fig.show()
```

```python
fig = px.line(df_2024.head(30), x="fecha", y="tipo_cambio", title="Tipo de cambio", markers=True)
fig.add_hline(y=df_2024["tipo_cambio"].head(30).mean())
fig.show()
```
