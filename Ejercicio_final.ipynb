### **¿Qué es PyCaret?**
PyCaret es una biblioteca de aprendizaje automático de código abierto y bajo código (*low-code*) en Python que automatiza los flujos de trabajo de Machine Learning. Permite pasar de la preparación de datos al despliegue del modelo en solo unas pocas líneas de código.


## **Paso 1: Instalación de PyCaret y Parche de Compatibilidad**

Instalamos la versión estable oficial de PyCaret y desactivamos la restricción artificial de versión de Python.


```python
# 1. Instalar la versión estable de PyCaret
!pip install pycaret

# 2. Parchear el archivo fuente de PyCaret para deshabilitar la restricción artificial de versión de Python 3.12+
import sys
import os

path = None
for p in sys.path:
    candidate = os.path.join(p, 'pycaret', '__init__.py')
    if os.path.exists(candidate):
        path = candidate
        break

if not path:
    path = '/usr/local/lib/python3.12/dist-packages/pycaret/__init__.py'

if os.path.exists(path):
    with open(path, 'r', encoding='utf-8') as f:
        content = f.read()

    # Cambiar la validación para que no bloquee Python 3.12
    content = content.replace('sys.version_info >= (3, 12)', 'sys.version_info >= (3, 99)')

    with open(path, 'w', encoding='utf-8') as f:
        f.write(content)
    print('¡Parche de compatibilidad aplicado con éxito en el archivo fuente de PyCaret!')
else:
    print('No se encontró el archivo de inicialización de PyCaret para parchar.')
```
