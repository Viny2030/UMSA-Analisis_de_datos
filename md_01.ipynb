# **<font color="#07a8ed">Markdown para cuadernos Google Colab**

<p align="justify">
✨ Markdown es un lenguaje de marcado ligero desarrollado por John Gruber en 2004, con la colaboración de Aaron Swartz. Fue diseñado con el objetivo de ser fácil de escribir y fácil de leer en su forma sin procesar, y convertirse en HTML de manera coherente. Desde su creación, Markdown ha ganado popularidad en todo el mundo debido a su simplicidad y versatilidad, convirtiéndose en uno de los lenguajes más utilizados en la web.
<br><br>
En cuanto a su uso en Google Colab, Markdown se integra como una de las opciones para escribir contenido en las celdas de texto. Google Colab es un entorno de Jupyter Notebook alojado por Google que permite a los usuarios escribir y ejecutar código Python en un navegador web. Además del código Python, los usuarios pueden incluir celdas de texto formateadas con Markdown para proporcionar explicaciones, documentación, comentarios y otros tipos de contenido.
<br><br>
El soporte de Markdown en Google Colab facilita la creación de documentos y notebooks interactivos, permitiendo a los usuarios combinar texto formateado con código ejecutable en un único documento. Esto es especialmente útil para la creación de tutoriales, informes de investigación, presentaciones y cualquier otro tipo de documento que requiera una combinación de texto y código.
<br><br>
👀 Los usuarios de Google Colab pueden utilizar Markdown para dar formato al texto, crear encabezados y títulos, crear listas y viñetas, incluir enlaces e imágenes, resaltar bloques de código y mucho más. Esta integración de Markdown hace que la experiencia de escribir y compartir contenido en Google Colab sea más fácil y flexible, permitiendo a los usuarios crear documentos más legibles y atractivos.
<br>

<p align="justify"><b>
En este lenguaje, particularmente en su sintaxis como se verá en las siguientes celdas de texto, se utilizan ciertos símbolos que le indican al lenguaje una acción específica, como por ejemplo en el caso de los títulos, depende del título que se quiere obtener, si se indica antes del texto, con el símbolo <code>#</code>.
<br><br>
Cuando se colocan esos símbolos especiales, el lenguaje interpreta que debe realizarse una acción determinada. Pero puede pasar que necesitemos utilizar en la escritura esos símbolos, y eliminar esa acción. Para ello, antes del símbolo utilizamos otro símbolo \ y se anula toda acción.
<br><br>

# **<font color="#07a8ed">Formatos de textos, imágenes y enlaces**

<p align="justify"><b>
Los títulos se definen agregando el símbolo <code>#</code> antes del texto, depende la cantidad de símbolos que se agreguen, será la jerarquía del título. Así para el título 1 se agrega un símbolo <code># Titulo 1</code>, para el titulo 2, dos símbolos <code>## Título 2</code>, y así sucesivamente. En algunos casos el título 5 y el título 6 tienen el mismo tamaño de letra, y depende del intérprete que se utilice, por ejemplo, si se usa JupyterLab los tipos de letras del título 5 y del título 6 son diferentes, y se usa Google Colab el título 5 y el título 6 son iguales.
<br><br>

**La sintaxis se debe escribir de la siguiente manera:**

~~~markdown
# Título 1
~~~

~~~markdown
## Título 2
~~~

~~~markdown
### Título 3
~~~

~~~markdown
#### Título 4
~~~

~~~markdown
##### Título 5
~~~

~~~markdown
###### Título 6
~~~



# **<font color="#07a8ed">Formatos de cursiva, negrita y de código**

## **<font color="#07a8ed">Cursiva**

<p align="justify"><b>
Para lograr el efecto del formato en cursiva, con un solo símbolo asterisco al inicio y final, de lo que se quiere con cursiva, se logra que la letra tenga ese aspecto. Es decir:
<br>

~~~markdown
*texto en cursiva*
~~~

👀 **y el resultado es el siguiente:**

*texto en cursiva*

<br>

## **<font color="#07a8ed">Negrita**

<p align="justify"><b>
El formato de negrita para el texto, se logra colocando antes de la palabra que se quiere en negrita dos veces el símbolo asterisco, y donde se quiere que finalice el formato de negrita, otros dos asteriscos sin espacio entre ellos. Es decir:
<br>

~~~markdown
**texto en negrita**
~~~

👀 **y el resultado es el siguiente:**

**texto en negrita**

<br>

## **<font color="#07a8ed">Cursiva y negrita**

<p align="justify"><b>
Ahora bien, para lograr la cursiva y la negrita simultáneamente, entonces se utilizan tres símbolos astericos como puede apreciarse, de la siguiente manera:
<br>

~~~markdown
***texto en cursiva y negrita***
~~~

👀 **y el resultado es el siguiente:**

***texto en cursiva y negrita***

<br>

## **<font color="#07a8ed">Formato de código**

👀 **Si se quiere en un párrafo dar el aspecto de código, en la primer línea se colocan los siguientes símbolos ~~~ y al final del párrafo también. La sintaxis es de la siguiente manera:**

\~~~

Este es mi párrafo de código

\~~~

👀 **y el resultado es el siguiente:**

~~~
Este es mi párrafo de código
~~~

<br><p>
👀 **Si se quiere que se respeten los colores de la sintaxis de acuerdo al lenguaje de código que se esta escribiendo, entonces se puede indicar tal lenguaje**

~~~python
pip install plotly
~~~

~~~latex
$n-1$
~~~

# **<font color="#07a8ed">Listas, numeración y listas anidadas**

## **<font color="#07a8ed">Listas**

👀 **Las listas se pueden crear con el símbolo `+`, con el símbolo `-`, o con el símbolo `\`, de la siguiente manera:**

~~~markdown
+ Primero se pone el símbolo,
+ Luego se escribe el texto,
+ y se repite la operación tanto como se necesite.
~~~

<br><p>
👀 **y el resultado es el siguiente:**

+ Primero se pone el símbolo,
+ Luego se escribe el texto,
+ y se repite la operación tanto como se necesite.

<br>

## **<font color="#07a8ed">Numeración**

<p align="jusfify"><b>
👀 Para la numeración, en vez de usar un símbolo, se puede colocar el número seguido de un punto, es decir 1. cada vez que se inicie una oración perteneciente a una numeración. Se puede colocar todas las veces 1., y cuando se ejecute la celda, se numerará como corresponde. Por ejemplo, se podría escribir lo siguiente:
<br>

~~~markdown
1. Coloco el 1. y empiezo a escribir,
1. Vuelvo a colocar 1., pero en la numeración se va a mostrar el 2.,
1. Repito la operación tantas veces como necesite numerar.
~~~

<br><p>
👀 **y el resultado es el siguiente:**

1. Coloco el 1. y empiezo a escribir,
1. Vuelvo a colocar 1., pero en la numeración se va a mostrar el 2.,
1. Repito la operación tantas veces como necesite numerar.

<br>

## **<font color="#07a8ed">Lista anidada**

<p align="justify"><b>
👀 Para generar una lista anidada, se debe dar cuatro espacios para incrementar el nivel de la lista, es decir:
<br>

~~~markdown
- Empiezo en el nivel uno,
    - Realizo cuatro espacios, para el nivel dos,
    - y continúo las veces necesarias,
    - si quiero incrementar el nivel, cuatro espacios mas,
        - y queda de esta forma,
        - nótese que automáticamente se cambia el símbolo de la lista.
~~~

<br><p>
👀 **y el resultado es el siguiente:**

- Empiezo en el nivel uno,
    - Realizo cuatro espacios, para el nivel dos,
    - y continúo las veces necesarias,
    - si quiero incrementar el nivel, cuatro espacios mas,
        - y queda de esta forma,
        - nótese que automáticamente se cambia el símbolo del la lista.

<br>

👀 **Lo mismo en la numeración:**

~~~markdown
1. Comienzo el nivel uno con 1.,
    1. doy cuatro espacios para el nivel dos,
    1. la numeración se reinicia para el nivel dos,
    1. concluido el nivel dos, podría subir de nivel, o volver al nivel 1.,
1. si vuelvo al nivel 1., debería continuar la numeración.
~~~

<br><p>
👀 **y el resultado es el siguiente:**

1. Comienzo el nivel uno con 1.,
    1. doy cuatro espacios para el nivel dos,
    1. la numeración se reinicia para el nivel dos,
    1. concluido el nivel dos, podría subir de nivel, o volver al nivel 1.,
1. si vuelvo al nivel 1., debería continuar la numeración.

<br>

👀 **Y también pueden combinarse, como el siguiente ejemplo:**

~~~markdown
1. Comienzo la numeración, y subo de nivel con una lista,
    - empiezo el nivel 2., con una lista,
    - continúo las veces necesarias,
1. vuelvo al nivel 1., numerado.
~~~

<br><p>
👀 **y el resultado es el siguiente:**

1. Comienzo la numeración, y subo de nivel con una lista,
    - empiezo el nivel 2., con una lista,
    - continúo las veces necesarias,
1. vuelvo al nivel 1., numerado.

<br>

# **<font color="#07a8ed">Enlaces e imágenes en línea**

## **<font color="#07a8ed">Enlaces**

<p align="justify"><b>
👀 Se pueden colocar directamente un enlace, por ejemplo, si quiero el enlace de Google, lo copio del navegador y lo pego a continuación. Quedaría de la siguiente forma:
<br>

https://www.google.com/

<br><p align="justify"><b>
👀 Si se quiere el enlace, pero referenciado a una palabra, es decir que al hacer clic en la palabra, nos lleve al enlace, entonces debe colocarse esa palabra entre corchetes, seguido del enlace entre paréntesis, como por ejemplo:
<br>

~~~markdown
[Google](https://www.google.com/)
~~~

<br><p>
👀 **y quedaría de la siguiente manera:**

[Google](https://www.google.com/)

<br>

## **<font color="#07a8ed">Imágenes en línea**

<p align="justify"><b>
Si se quiere obtener una imágen en línea, como por ejemplo el logo de Google, se debe escribir el código de la misma manera que el enlace referenciado a una palabra, pero iniciando el código con el signo de exclamación, y además el enlace debe estar completo, indicando la ubicación de la imágen que se quiere traer a la celda de texto, por ejemplo:
<br>

~~~markdown
![Google](https://www.google.com/images/logos/google_logo_41.png)
~~~

<br><p>
**y quedaría de la siguiente manera:**

![Google](https://www.google.com/images/logos/google_logo_41.png)
