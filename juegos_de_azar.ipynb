 # **<font color="DarkBlue">Loto5</font>**


🚀 Loto5 es un juego donde se juegan 5 numeros del 0 al 36

```python
import random  #### loto 5
def unico(x,L):
  esUnico=True
  for i in range(len(L)):
    if x==L[i]:
      esUnico=False
      break
  return esUnico
L=[]
j=0
while j<5:
  x=random.randint(0,36)
  if unico(x,L):
    L.append(x)
    j+=1

L.sort()
print(L)
```
