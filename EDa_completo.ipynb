### Planteo del Problema de Negocio

**Problema de Regresión: Predicción de Precio**
El objetivo principal de este conjunto de datos es predecir el precio sugerido de venta al público (Precio_Sugerido) de un vehículo basándose en sus características.

**Tipo de Aprendizaje: Supervisado.**

**Variable Objetivo (Target): Precio_Sugerido (MSRP)**. Esta es una variable continua (de tipo int64 y con un amplio rango de valores), lo que define el problema como de Regresión.

Características (Features): Todas las demás columnas, como Fabricante, Modelo, Potencia_CV, MPG_Carretera, Tipo_Transmision, etc.

**Formulación del Problema**
Se busca construir un modelo que pueda estimar con la mayor precisión posible el valor de Precio_Sugerido para cualquier coche nuevo o no visto en el conjunto de entrenamiento, basándose en un conjunto de características técnicas y de mercado.
