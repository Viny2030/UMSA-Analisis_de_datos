1. .T: Transpone inmediatamente cada DataFrame. Esto hace que las claves del diccionario (ventas, costo, bruto) se conviertan en el índice de las filas y los índices de las Series (los años) se conviertan en las columnas.

 # **<font color="#07a8ed">Correlación y covarianza</font>**

<p align="justify">
Algunas estadísticas de resumen, como la correlación y la covarianza, se calculan a partir de argumentos. </p>

```python
resultado = sucursal1.T
resultado
```

Transpone el DataFrame sucursal1 y lo asigna a una nueva variable resultado.



<p align="justify">
Algunas estadísticas de resumen, como la correlación y la covarianza, se calculan a partir de argumentos. </p>

<p align="justify">
👀 Podemos calcular la correlación con el método <code>corr()</code> entre las ventas y los costos.
</p>

```python
resultado["ventas"].corr(resultado["costo"]).round(2)
```

```python
resultado.ventas.corr(resultado.costo).round(2)
```

Muestran dos formas de calcular la correlación entre dos columnas ("ventas" y "costo"). Se puede acceder a las columnas con la sintaxis de corchetes ["columna"] o, si el nombre no tiene espacios, con la sintaxis de punto .columna.



<p align="justify">
La correlación es una medida estadística que indica la fuerza y la dirección de la relación entre dos variables. Se mide con el coeficiente de correlación, que puede variar entre -1 y 1:
<br><br>

- **1** indica una correlación positiva perfecta: cuando una variable aumenta, la otra también lo hace en proporción directa.
- **-1** indica una correlación negativa perfecta: cuando una variable aumenta, la otra disminuye en proporción directa.
- **0** indica que no hay correlación lineal entre las variables.

<p align="justify">
<br>
En el script <code>resultado["ventas"].corr(resultado["costo"])</code>, se está calculando el coeficiente de correlación entre dos columnas de un DataFrame de pandas: <code>ventas</code> y <code>costo</code>.
<br><br>

- **`resultado["ventas"]`**: Es una serie que representa las ventas en un conjunto de datos.
- **`resultado["costo"]`**: Es una serie que representa los costos en el mismo conjunto de datos.

<p align="justify">
<br>
La función <code>corr()</code> calcula el coeficiente de correlación entre estas dos series. El valor resultante te dirá cómo están relacionadas las ventas y los costos:
<br><br>

- Si el coeficiente es cercano a 1, hay una fuerte relación positiva: a mayor costo, mayores ventas.
- Si el coeficiente es cercano a -1, hay una fuerte relación negativa: a mayor costo, menores ventas.
- Si el coeficiente es cercano a 0, no hay una relación lineal significativa entre ventas y costos.



<p align="justify">
👀 También podemos calcular la covarianza con el método <code>cov()</code> entre las ventas y los costos.
</p>

```python
resultado["ventas"].cov(resultado["costo"]).round(2)
```

```python
resultado.ventas.cov(resultado.costo).round(2)
```

De manera similar, calculan la covarianza entre las mismas dos columnas.



<p align="justify">
La <code>covarianza</code> es una medida que indica la dirección de la relación lineal entre dos variables. A diferencia de la correlación, la covarianza no está normalizada y puede tomar cualquier valor, por lo que su interpretación no es tan intuitiva.
<br><br>

- **Covarianza positiva**: Cuando una variable aumenta, la otra también tiende a aumentar. Indica una relación directa entre las dos variables.
- **Covarianza negativa**: Cuando una variable aumenta, la otra tiende a disminuir. Indica una relación inversa entre las dos variables.
- **Covarianza cercana a cero**: Indica que no hay una relación lineal clara entre las variables.

<p align="justify">
<br>
En el script <code>resultado["ventas"].cov(resultado["costo"]).round(2)</code>, se está calculando la covarianza entre dos columnas del DataFrame de pandas: <code>ventas</code> y <code>costo</code>.

- **`resultado["ventas"]`**: Es una serie que representa las ventas en un conjunto de datos.
- **`resultado["costo"]`**: Es una serie que representa los costos en el mismo conjunto de datos.

<p align="justify">
<br>
El valor resultante te dirá cómo varían conjuntamente las ventas y los costos:

- **Valor positivo**: Indica que, en general, cuando los costos aumentan, también lo hacen las ventas, y viceversa.
- **Valor negativo**: Indica que cuando los costos aumentan, las ventas tienden a disminuir, y viceversa.
- **Valor cercano a cero**: Indica que no hay una relación lineal fuerte entre ventas y costos.

<p align="justify">
<br>
La covarianza proporciona una idea de cómo las dos variables cambian juntas, pero a diferencia de la correlación, no indica la fuerza de la relación ni está normalizada.

<p align="justify">

</p>

Al llamar al método .corr() en un DataFrame completo, Pandas calcula la matriz de correlación, que muestra la correlación entre cada par de columnas del DataFrame.
