### 📊 Análisis de Matriz de Dispersión (Vista Global y Estructura)

**Observación principal:** Esta visualización analiza la distribución general de las variables. La inclusión de la columna índice (`Unnamed: 0`) expone patrones de ordenamiento en el dataset que deben corregirse antes del modelado.

| 📌 Variable / Relación | 📈 Patrón Visual Observado | 💡 Insight Analítico para Preprocesamiento |
| :--- | :--- | :--- |
| **`Unnamed: 0` (Índice)** | Histograma plano. En dispersión, genera<br>"bloques" densos y espacios en blanco notorios. | 🚨 **¡Alerta de ordenamiento!** El dataset no está barajado<br>aleatoriamente. Los datos están agrupados en bloques.<br>**Esta columna debe eliminarse antes de modelar**. |
| **`ultimaevaluacion`** | Histograma de barras continuas.<br>Distribución relativamente plana y amplia. | Al no estar segmentada por abandono, se ve una<br>masa uniforme que oculta el comportamiento de los extremos. |
| **`numerosdeproyectos`** | Histograma discreto. Concentración masiva<br>en los valores de 3 y 4 proyectos. | La carga operativa estándar de la empresa está<br>centralizada. Los casos de 2 y 6-7 proyectos son la minoría. |
| **`horasmensualespromedio`** | Histograma amplio (de 100 a >300 hs).<br>Ligeramente multimodal (varios picos suaves). | Existe una dispersión enorme en la carga horaria global,<br>lo que sugiere alta disparidad de roles o mala gestión. |
| **`tiempoenlaempresa`** | Fuerte asimetría a la derecha. Concentración<br>masiva en los años 2, 3 y 4. | La plantilla es mayoritariamente "nueva". Hay una carencia<br>evidente de retención de talento a largo plazo (> 6 años). |
| **`Unnamed: 0` vs Resto** | Franjas densas al inicio (0-2000) y al final<br>(11000-15000), con huecos claros en el medio. | Confirma que el archivo CSV original fue ordenado por<br>alguna variable (muy probablemente la de abandono). |
| **`proyectos` vs `horas`** | Frontera inferior en forma de escalera.<br>Espacio vacío en la esquina inferior derecha. | Límite estructural: a mayor cantidad de proyectos, el piso<br>mínimo de horas sube mecánicamente. |

> **🛠️ Acción recomendada en código:** > 1. Ejecutar `df = df.drop(columns=['Unnamed: 0'])` para evitar fuga de información (*data leakage*).
> 2. Ejecutar `df = df.sample(frac=1, random_state=42).reset_index(drop=True)` para mezclar los datos (*shuffling*) y romper el sesgo de ordenamiento que expone el gráfico.

### 📊 Análisis de Matriz de Dispersión: Patrones de Atrición

**Leyenda:**
* 🔵 **Clase 0:** Retención (Continúan en la empresa)
* 🟠 **Clase 1:** Abandono (Dejaron la empresa)

| 📌 Variable / Relación | 🔵 Patrón de Retención | 🟠 Patrón de Abandono | 💡 Insight Analítico |
| :--- | :--- | :--- | :--- |
| **`ultimaevaluacion`** | Distribución normal (campana)<br>centrada alrededor de 0.7. | **Bimodal:** Picos marcados<br>en < 0.6 y > 0.8. | Se pierde talento en ambos extremos:<br>empleados de bajo rendimiento y "estrellas". |
| **`numerosdeproyectos`** | Fuerte concentración en la<br>zona media (3, 4 y 5 proyectos). | Concentración en los extremos:<br>exactamente **2** y de **6 a 7** proyectos. | Identifica dos fallas de gestión:<br>la **subutilización** y el **burnout** o sobrecarga. |
| **`horasmensualespromedio`** | Distribución unimodal, agrupada<br>en ritmo de trabajo estándar. | **Bimodal:** Un grupo con ~150 hs<br>y otro que supera las 250 hs. | El exceso crónico de horas<br>es un fuerte predictor de salida. |
| **`tiempoenlaempresa`** | Mayor volumen a los 3 años,<br>con decaimiento gradual. | Ventana crítica y picos<br>de salida en los años **3, 4 y 5**. | Existe un umbral de fidelización:<br>quienes superan los 6 años rara vez se van. |
| **`evaluacion` vs `horas`** | Nube de dispersión uniforme<br>en el centro del gráfico. | **Dos clústeres aislados:** <br>1. Baja evaluación + Pocas horas.<br>2. Alta evaluación + Muchas horas. | Confirma dos perfiles distintos de salida:<br>el empleado desmotivado y el "estrella" agotado. |
| **`proyectos` vs `horas`** | Carga horaria equilibrada<br>para 3-5 proyectos (150-250 hs). | **Correlación extrema:** Quienes tienen<br>6-7 proyectos siempre trabajan > 250 hs. | Asignar más de 5 proyectos empuja mecánicamente<br>al exceso de horas y al posterior abandono. |

> **Nota para modelado:** La fuerte presencia de distribuciones bimodales y clústeres no lineales sugiere que algoritmos basados en árboles (Random Forest, Gradient Boosting) tendrán mejor rendimiento que regresiones logísticas tradicionales.
